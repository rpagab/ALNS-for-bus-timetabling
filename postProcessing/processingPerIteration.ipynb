```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import itertools
import os
import statistics as stat
colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k']
color_cycle = itertools.cycle(colors)
```

```python
def add_variables(dataRaw, ops):
    """Compute and append derived operator statistics to a per-iteration DataFrame.

    Parameters
    ----------
    dataRaw : pandas.DataFrame
        Raw DataFrame containing per-iteration operator statistics. Expected to
        contain columns like ``duration <i>``, ``nb uses <i>``, ``avgDuration <i>``,
        ``proba <i>``, ``sigmas1 <i>``, ``sigmas2 <i>``, ``sigmas6 <i>`` for each
        operator id in ``ops``.
    ops : iterable
        Iterable of operator ids (integers) for which derived columns
        should be computed.

    Returns
    -------
    pandas.DataFrame
        The input ``dataRaw`` DataFrame with additional columns added for totals,
        shares, expected durations, improvements, scores per time, infeasibility
        counts, and other derived metrics (column names include the operator id
        as a suffix, e.g., ``"share duration 1"``).

    Notes
    -----
    Logic: sums operator durations and uses to compute totals, then for each
    operator computes ratios (shares), expected iteration duration, improvements
    and normalized metrics. Also computes time-based differences (e.g.,
    ``duration op last iter``) and handles NaNs by filling missing values where
    appropriate.
    """
    dataRaw["tot duration"] = sum([dataRaw["duration "+str(i)] for i in ops])
    dataRaw["tot nb uses"] = sum([dataRaw["nb uses "+str(i)] for i in ops])
    dataRaw["tot avgDuration"]=dataRaw["tot duration"]/dataRaw["tot nb uses"]
    
    for i in ops:
        dataRaw["share duration "+str(i)]=dataRaw["duration "+str(i)]/dataRaw["tot duration"]
        dataRaw["expected iter duration "+str(i)]=(dataRaw["proba "+str(i)] * dataRaw["avgDuration "+str(i)]).fillna(0)
        dataRaw["duration except "+str(i)]=dataRaw["tot duration"]-dataRaw["duration "+str(i)]

        dataRaw["improvements "+str(i)]=dataRaw["sigmas1 "+str(i)]+dataRaw["sigmas2 "+str(i)]
        dataRaw["share improvements "+str(i)]=dataRaw["improvements "+str(i)]/dataRaw["nb uses "+str(i)]
        dataRaw["improvements per time "+str(i)]=dataRaw["improvements "+str(i)]/dataRaw["duration "+str(i)]
        dataRaw["score per time "+str(i)]=dataRaw["score "+str(i)]/dataRaw["avgDuration "+str(i)]

        dataRaw["op infeasible "+str(i)]=dataRaw["sigmas6 "+str(i)]
        dataRaw["share op infeasible "+str(i)]=dataRaw["op infeasible "+str(i)]/dataRaw["nb uses "+str(i)]
        dataRaw["duration op last iter "+str(i)]=dataRaw["duration "+str(i)].diff()
        
    for i in ops:
        dataRaw["expected share duration "+str(i)]=dataRaw["expected iter duration "+str(i)]/dataRaw[["expected iter duration "+str(j) for j in ops]].fillna(0).sum(axis=1)

    return dataRaw
    
```

```python
def print_graph(folder, instance, runs, params, minTime, maxTime, variablesGen, variablesOp, operators):
    """Plot evolution graphs for a given instance across runs and parameter sets.

    Parameters
    ----------
    folder : str
        Path to the folder containing evolution files. Files are expected to follow
        the naming pattern: ``Evo_<instance>_<param0>_<run>.txt`` where ``param0`` is
        ``p[0]`` from the entries in ``params``.
    instance : str
        Instance identifier (e.g., 'L1').
    runs : iterable
        Iterable of run indices (integers) to include in the plots.
    params : iterable
        Iterable of parameter descriptors; each element ``p`` should provide
        ``p[0]`` (filename token) and ``p[1]`` (label token used in legend).
    minTime, maxTime : numeric
        Time window (inclusive) used to select rows for plotting.
    variablesGen : iterable
        Iterable of tuples ``(axisFlag, columnName)`` for general variables to
        plot. ``axisFlag``==1 plots on the primary y-axis, ``==2`` on the
        secondary y-axis.
    variablesOp : iterable
        Iterable of tuples ``(axisFlag, columnName)`` for operator-specific
        columns; actual column names are formed by appending ``" " + str(i)``
        for each operator id in ``operators``.
    operators : iterable
        Iterable of operator ids (integers) to expand operator-specific columns.

    Returns
    -------
    None
        Displays the matplotlib figure; the function does not return a value.

    Notes
    -----
    Logic: for each run and parameter tuple, the function reads the corresponding
    evolution file into a DataFrame, augments it with ``add_variables``, filters
    by the requested time window, and plots the requested columns on two
    y-axes with labels and legends.
    """
    fig, ax1 = plt.subplots()
    ax2 = ax1.twinx()
    fig.set_size_inches(15, 6.5)

    for r in runs:
        for p in params:
            inputFile = folder + "/Evo_" + instance + "_" + p[0] + "_" + str(r) + ".txt"
            dataRaw = pd.read_csv(inputFile, sep='\t', index_col=False)
            data = add_variables(dataRaw, [1, 2, 3, 4, 5])
            dataPeriod = data[(data["time"].between(minTime, maxTime))]

            for vG in variablesGen:
                if vG[0] == 1:
                    dataPeriod.plot(x="time", y=vG[1], ax = ax1, marker='x', markersize = 5, legend=True, label = vG[1]+"_"+p[1], ylabel = vG[1], color=next(color_cycle))
                if vG[0] == 2:
                    dataPeriod.plot(x="time", y=vG[1], ax = ax2, marker='+', linestyle='--', markersize = 5, legend=True, label = vG[1]+"_"+p[1], ylabel = vG[1], color=next(color_cycle))

            for vO in variablesOp:
                if vO[0] == 1:
                    for i in operators:
                        dataPeriod.plot(x="time", y=vO[1]+" "+str(i), ax = ax1, marker='x', markersize = 5, label = vO[1]+"_"+str(i)+"_"+p[1], ylabel = vO[1])
                if vO[0] == 2:
                    for i in operators:
                        dataPeriod.plot(x="time", y=vO[1]+" "+str(i), ax = ax2, linestyle='--', marker='+', markersize = 5, label = vO[1]+"_"+str(i)+"_"+p[1], ylabel = vO[1])

    plt.xlim(left=minTime)
    ax1.set_xlabel("Time (seconds)")
    plt.show()
```

# 2. Average computation time of each operator (section 5.3, Table 3)

```python

heurOp = {
    "name": "Heuristic operator",
    "folder": "../results/impactOperators/removingMILPOperator/PerIteration/",
    "params": "7200_8_10_0.1_1_0.005_1e-8_100_1e-4_2_0",
    "operator": 1
}

MILPOp = {
    "name": "MILP operator",
    "folder": "../results/impactOperators/removingHeuristicOperator/PerIteration/",
    "params": "7200_8_01_0.1_1_0.005_1e-8_100_1e-4_2_0",
    "operator": 5
}

instances = [
"S1", "S2", "S3",
"M1", "M2", "M3",
"L1", "L2", "L3"
]
for op in (heurOp, MILPOp):
    print("\nResults for:", op["name"])
    folder = op["folder"]
    params = op["params"]
    operator = op["operator"]
    inputFiles = os.listdir(folder)
    index = pd.Index(range(1, 11), name="runs")

    dfAvgDurations = pd.DataFrame([[None for _ in range(len(instances))] for _ in range(1, 11)], index = index, columns = instances)

    for inputFile in inputFiles:
        if params in inputFile:
            for inst in instances:
                if inputFile.startswith("Evo_" + inst):
                    dataRaw = pd.read_csv(folder + inputFile, sep='\t', index_col=False)
                    run = int(inputFile.split("_")[-1].split(".")[0])
                    dfAvgDurations.loc[run, inst] = dataRaw["avgDuration " + str(operator)].iloc[-1]

    for size in ("S", "M", "L"):
        avgDurations = []
        for i in range(1, 4):
            inst = size + str(i)
            avgDurations += dfAvgDurations[inst].dropna().tolist()
        print(f"Size {size}:")
        print(f"  Average computation time: {stat.mean(avgDurations):.5f} seconds")
```
