```python
inputFile = "../results/baseALNS/Timetable/Tim_S3_7200_8_11_0.1_1_0.005_1e-8_100_1e-4_2_0_1.txt"  # the timetable to be reformatted
outputFile = "examplesFormattedTimetables/Tim_S3_7200_8_11_0.1_1_0.005_1e-8_100_1e-4_2_0_1.txt"
with open(inputFile, 'r') as inputFile:
    with open(outputFile, 'w') as outputFile:
        lines = inputFile.readlines()
        # write the header
        outputFile.write(lines[0])
        # write the content
        for i, inputline in enumerate(lines[1:]):
            parts = inputline.strip().split(sep = '\t')
            if i % 2 == 0:  # arrival time
                times = [f"{(int(m) + 7 * 60) // 60:02d}:{(int(m) + 7 * 60) % 60:02d}" for m in parts[4:]]
                outputLine = '\t'.join(parts[:3] + times) + '\n'
            else:  # departure time
                times = [f"{(int(m) + 7 * 60) // 60:02d}:{(int(m) + 7 * 60) % 60:02d}" for m in parts[4:-1]]
                outputLine = '\t'.join(parts[:3] + times) + '\t/\n'

            outputFile.write(outputLine)
        
```
